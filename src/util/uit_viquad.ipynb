```python
import json
from datasets import load_dataset

# Load UIT-ViQuAD from Hugging Face
dataset = load_dataset("taidng/UIT-ViQuAD2.0")

# Select the split you want
split = "validation"

# Store contexts and their associated QAs
context_map = {}

for example in dataset[split]:
    context = example["context"]

    # Create a new context entry if it doesn't exist
    if context not in context_map:
        context_map[context] = {
            "context": context,
            "qas": []
        }

    # Extract answers
    answers = example.get("answers", {})

    answer_texts = answers.get("text", [])
    answer_starts = answers.get("answer_start", [])

    # Construct SQuAD-style answers
    formatted_answers = [
        {
            "text": text,
            "answer_start": start
        }
        for text, start in zip(answer_texts, answer_starts)
    ]

    # Add the QA pair
    context_map[context]["qas"].append({
        "id": example["id"],
        "question": example["question"],
        "answers": formatted_answers,
        "is_impossible": example.get("is_impossible", False)
    })

# Construct final SQuAD-style dataset
output = {
    "data": [
        {
            "paragraphs": list(context_map.values())
        }
    ]
}

# Save to JSON
output_file = r"data\results\uit_viquad_val_squad.json"

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(output, f, ensure_ascii=False, indent=2)

print(f"Saved to: {output_file}")
print(f"Number of unique contexts: {len(context_map)}")
print(f"Number of QA pairs: {len(dataset[split])}")
```
