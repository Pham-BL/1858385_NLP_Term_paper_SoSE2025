```python
import json
import csv

# Read JSON
with open(r"\data\rq1\context_mapping_phoasr_VIQUAD_val.json", "r", encoding="utf-8") as f: #swap between phoasr train and test file and visqa test file
    data = json.load(f)

# Write CSV
with open("\data\rq1\ec\input\output_phoasr_test.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["asr_text", "reference_text"]
    )

    writer.writeheader()

    for item in data["results"]:
        writer.writerow({
            "asr_text": item.get("transcription", ""),
            "reference_text": item.get("matched_context", "")
        })
```
