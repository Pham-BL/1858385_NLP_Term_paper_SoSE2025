```python
import json
import pathlib as Path

file1 = Path(r"data\output\rq1\ft_vit5_visqa_test_results.json") #Swap out the files accordingly, see readme.txt
file2 = Path(r"data\input\rq2\wer_calc_VISQA_VIQUAD_val.json")
output_file = Path(r"data\input\rq2\ft_VISQA_VIQUAD_val.json")

# Load file 1
with open(file1, "r", encoding="utf-8") as f:
    data1 = json.load(f)

# Load file 2
with open(file2, "r", encoding="utf-8") as f:
    data2 = json.load(f)

# Replace transcription in file 2 with vit5_prediction from file 1
for i, result2 in enumerate(data2["results"]):
    result2["transcription"] = data1["results"][i]["vit5_prediction"]

# Save as a new file
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(data2, f, ensure_ascii=False, indent=2)

print(f"Saved updated file to: {output_file}")
```
